## 1. Python, programs, and the interpreter

**Python is a programming language.** We use it to write instructions for a computer. A **program** is a set of instructions.

The **Python interpreter** is the software that runs Python code. If we write `print(2 + 3)`, Python calculates the sum and shows `5`. We use Python in AI to prepare data, call models, and build applications.



Python code is run by the interpreter. The result appears as output.

## 2. Jupyter Notebook, cells, and the kernel

A **Jupyter Notebook** is a document that contains text, code, and saved results. Its file name ends with `.ipynb`.

| Term | Meaning |
|---|---|
| Markdown cell | A cell for headings, explanations, tables, and images |
| Code cell | A cell containing code that Python can run |
| Output | The result shown below a code cell |

This is our first code cell. `print()` is a **built-in function**, a ready-made action supplied by Python. It displays the value between its parentheses. Text goes inside quotes.

```python
print('Hello, Python!')  # Display a text message.
```

Python can display a number too. A number does not need quotes.

```python
print(25)  # Display a whole number.
```

**Syntax** means the rules for writing code. For example, quote marks around text must match.

A **statement** is an instruction Python runs. An **expression** is code that produces a value. `2 + 3` is an expression with the value `5`.

A **comment** explains code to a person. Python ignores text after `#` on that line. Comments help us remember why a line is needed.

```python
print(2 + 3)  # Calculate the expression, then display its value.
```

## 4. Variables and assignment

A **variable** is a name that refers to a value in memory. We create it with **assignment**: `name = value`.

In `score = 80`, `score` is the variable name, `=` is the assignment operator, and `80` is the value. Python evaluates the right side first, then binds the name on the left to that result. We use variables so we can reuse and change values.

```python
score = 80  # Assign the value 80 to the name score.
```

Assignment does not display a result. Use the variable name with `print()` to see its value.

```python
print(score)
```

**Reassignment** gives the same name a new value. The next cell replaces the value that `score` refers to.

```python
score = 85  # Update the score.
```

```python
print(score)
```

The right side can be an expression. We use the current score to calculate its new value.

```python
score = score + 5  # Read 85, add 5, then assign 90.
```

```python
print(score)
```

## 5. Objects and data types

An **object** is a value that Python keeps in memory. Each object has a **type** and a **value**. For example, the object `80` has type `int` and value `80`.

The variable is the name that refers to the object. We only need this meaning of object here; we do not need to create classes.



score is a variable name. It refers to an integer object whose value is 80.

A **data type** tells us what kind of value an object holds and what operations it supports.

| Type | Name | Example |
|---|---|---|
| `int` | Integer, a whole number | `12` |
| `float` | Floating-point number, often written with a decimal point | `12.5` |
| `bool` | Boolean, a true or false value | `True` |
| `str` | String, a sequence of text characters | `"Python"` |
| `NoneType` | The type of `None`, meaning no value is present | `None` |

```python
age = 18  # int: a whole number.
```

```python
price = 12.5  # float: a decimal number.
```

```python
is_open = True  # bool: True and False begin with capital letters.
```

```python
course = 'Python'  # str: text inside quotes.
```

```python
result = None  # No result has been set yet.
```

The built-in function `type()` returns the type of an object. We call it to check what Python is holding.

```python
type(age)
```

```python
type(price)
```

```python
type(is_open)
```

```python
type(course)
```

```python
type(result)
```

Jupyter displays the value of the last expression in a cell. That is why `type(age)` shows a result without `print()`. We can also write `print(type(age))`.

## 6. Identifiers and keywords

An **identifier** is a name used for a variable or function. A **keyword** is a word reserved by Python, such as `if`, `for`, or `def`.

| Valid names | Invalid names | Reason |
|---|---|---|
| `student_name` | `student name` | A name cannot contain a space |
| `score2` | `2score` | A name cannot start with a digit |
| `total_price` | `total-price` | `-` is an operator, not part of a name |
| `class_name` | `class` | `class` is a Python keyword |

Names are **case-sensitive**: `score` and `Score` are different. Use clear names with underscores. Avoid names such as `print`, `str`, or `list`, because they hide Python functions we need.

```python
student_name = 'Asha'  # A clear identifier.
```

```python
print(student_name)
```

## 7. Type conversion

**Type conversion**, also called **type casting**, changes a value into another type when that change is valid.

`"12"` is text, while `12` is a number. We convert text to a number before doing number calculations. `int()` makes an integer, `float()` makes a floating-point number, and `str()` makes text.

```python
quantity_text = '3'  # Quotes make this a string.
```

```python
type(quantity_text)
```

```python
quantity = int(quantity_text)  # Convert valid whole-number text.
```

```python
print(quantity + 2)
```

```python
float('3.5')  # Convert decimal text to a float.
```

```python
str(3) + ' items'  # Convert the number before joining it to text.
```

`int("three")` cannot work because the word is not written as a number. `int(3.9)` gives `3`; it removes the fractional part toward zero, rather than rounding. In mixed arithmetic such as `2 + 0.5`, Python uses a float for the result.

```python
2 + 0.5  # An integer and a float produce a float here.
```

## 8. Strings and string indexing

A **string** is an ordered sequence of characters. A character can be a letter, digit, space, or punctuation mark.

**Indexing** selects one character by its position. The first index is `0`. A negative index counts from the end, so `-1` selects the last character.

```python
word = 'PYTHON'
```

![Indexes begin at 0. The orange cells are the characters selected by word[1:4].](attachment:string_index.png)

Indexes begin at 0. The orange cells are the characters selected by word[1:4].

```python
word[0]  # The first character: P.
```

```python
word[2]  # The third character: T.
```

```python
word[-1]  # The last character: N.
```

```python
len(word)  # len() counts the six characters.
```

## 9. String slicing

**Slicing** selects part of a sequence. The syntax is `sequence[start:stop:step]`.

| Part | Meaning |
|---|---|
| `start` | First index to include |
| `stop` | Index to stop before |
| `step` | How far to move each time; the default is 1 |

```python
word[1:4]  # Include 1, 2, and 3. Exclude 4.
```

```python
word[:3]  # Start at the beginning.
```

```python
word[3:]  # Continue to the end.
```

```python
word[::2]  # Take every second character.
```

## 10. String operations and methods

**Concatenation** joins strings. The `+` operator joins two strings, and `*` repeats a string.

A **method** is a function associated with an object. We call it using a dot, as in `word.lower()`. The parentheses mean we are calling the method.

```python
'Hello' + ' ' + 'Asha'  # Join three strings, including a space.
```

```python
'ha' * 3  # Repeat the string three times.
```

The `lower()` method returns text with lowercase letters. `upper()` returns uppercase letters. These methods are useful when data uses mixed letter cases.

```python
word.lower()
```

```python
'Python'.upper()
```

The `strip()` method removes spaces at the beginning and end. It does not remove spaces between words.

```python
'  Asha Khan  '.strip()
```

The `replace(old, new)` method returns text with a chosen part replaced. We give it the old text first and the new text second.

```python
'I like tea'.replace('tea', 'milk')
```

The `split(separator)` method cuts text into parts and returns a list. We will study lists in Part 2.

```python
'red,green,blue'.split(',')
```

Strings are **immutable**: their characters cannot be changed in place. A method such as `lower()` returns a new string. The original string stays the same unless we assign the new result.

```python
print(word)  # The original value is still PYTHON.
```

```python
word = word.lower()  # Now the name refers to the new lowercase string.
```

```python
print(word)
```

## 11. Formatted strings: f-strings

An **f-string** inserts values into text. Write `f` before the quote and put an expression inside `{}`. We use it to build a sentence from variables.

```python
name = 'Asha'
```

```python
marks = 85
```



The values of name and marks are inserted where their names appear inside braces.

```python
print(f'{name} scored {marks} marks.')
```
