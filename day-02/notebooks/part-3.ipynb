## 1. Loops and iterations

A **loop** repeats a block of code. One run of that block is an **iteration**.

Use a `for` loop to visit items in a sequence. Use a `while` loop to repeat while a condition remains true. For example, a shopping list can have many items, but we can print all of them with one loop.

## 2. The for loop

A **for loop** takes one item at a time from an **iterable**, an object we can get items from, such as a list, string, or range.

In `for fruit in fruits:`, `fruit` is the **loop variable**. Python assigns the next item to it before running the indented block. The colon starts the block.

```python
fruits = ['apple', 'banana', 'mango']
```

```python
for fruit in fruits:
    print(fruit)  # The same instruction runs for each item.
```



The loop runs three iterations. The loop variable receives one item each time.

A string is iterable too. Each iteration gives us one character.

```python
for letter in 'AI':
    print(letter)
```

## 3. The range() function

`range()` represents an integer sequence. It is useful when a loop needs to run a known number of times.

| Form | Meaning |
|---|---|
| `range(stop)` | Start at 0 and stop before `stop` |
| `range(start, stop)` | Start at `start` and stop before `stop` |
| `range(start, stop, step)` | Move by `step` each time |

The stop value is always excluded. `range()` does not create a list of all the numbers in memory.

```python
list(range(3))  # Convert to a list so we can inspect 0, 1, 2.
```

```python
for day in range(1, 4):
    print(day)  # Print 1, 2, and 3.
```

```python
list(range(0, 7, 2))  # Start at 0, move by 2, stop before 7.
```

## 4. Accumulation and enumerate()

An **accumulator** is a variable that keeps a running result. We start a total at 0, then add one value in each iteration.

```python
prices = [10, 20, 30]
```

```python
total = 0  # Start with no items counted.
```

```python
for price in prices:
    total += price  # Add the current item to the running total.
```

```python
print(total)
```

Trace the values: before the loop, `total` is 0. After each iteration it becomes 10, then 30, then 60.

`enumerate()` gives an index and an item together. We use `start=1` when we want human-readable list numbers starting at 1.

```python
for number, fruit in enumerate(fruits, start=1):
    print(number, fruit)
```

## 5. The while loop

A **while loop** checks a condition before every iteration. If the condition is False, it stops.

A counter loop needs three parts: an initial value, a condition, and an update. If the update is missing, the loop may never stop. This is called an **infinite loop**.



The counter starts at 1. After printing 3, it becomes 4, so count <= 3 is False.

```python
count = 1  # Initial value.
```

```python
while count <= 3:  # Condition.
    print(count)
    count += 1  # Update.
```

```python
print(count)  # The value is 4 when the loop ends.
```

## 6. break and continue

`break` leaves the loop immediately. `continue` skips the rest of the current iteration and starts the next one.

Use `break` when a search has found what it needs. Use `continue` when one item should be skipped.

```python
for fruit in fruits:
    if fruit == 'banana':
        print('Found banana')
        break
```

```python
for number in range(1, 5):
    if number == 2:
        continue  # Skip 2 only.
    print(number)
```

## 7. Functions: definition and function call

A **function** is a reusable block of code that performs a task. We write its steps once and **call** the function whenever we need that task.

| Term | Meaning |
|---|---|
| Function definition | The code that creates the function |
| `def` | The keyword used to define a function |
| Function name | The name used to call it |
| Function body | The indented instructions inside it |
| Function call | Running the function using parentheses |

Defining a function does not run its body.

```python
def greet():
    print('Hello, learner!')  # Function body.
```

```python
greet()  # First function call.
```

```python
greet()  # Reuse the same function.
```

## 8. Parameters and arguments

A **parameter** is a variable name in a function definition. An **argument** is an actual value supplied in a function call.

In `def double(number):`, `number` is a parameter. In `double(4)`, `4` is an argument. The argument is assigned to the parameter when the function starts.



The argument 4 becomes the value of the parameter number. The returned value is 8.

```python
def double(number):
    return number * 2  # Send the calculated value back.
```

```python
double(4)
```

```python
double(7)  # The same function works with a different argument.
```

## 9. Return values and print()

A **return value** is the result a function sends back to its caller. `return` also ends that function call. We can store the returned value and use it in another calculation.

`print()` displays text or values. Displaying a value does not make it the function's return value. A function without an explicit `return` returns `None`.

```python
answer = double(5)  # Store the return value 10.
```

```python
answer + 3  # Use it in another expression.
```

```python
def show_double(number):
    print(number * 2)  # Display a value, but do not return it.
```

```python
shown_result = show_double(5)  # Displays 10.
```

```python
print(shown_result)  # None: no return statement was used.
```

## 10. Multiple parameters and keyword arguments

A function can receive more than one parameter. **Positional arguments** match parameters by order. **Keyword arguments** name the parameter explicitly.

We will calculate a rectangle's area. The parameters are `length` and `width`, and the return value is their product.

```python
def rectangle_area(length, width):
    return length * width
```

```python
rectangle_area(5, 3)  # Positional arguments: length=5, width=3.
```

```python
rectangle_area(width=3, length=5)  # Names make the meaning clear.
```

## 11. Default parameters

A **default value** is used when a call leaves out an argument. We put the default in the function definition.

Our greeting uses `"learner"` when no name is supplied. We can still supply another name.

```python
def welcome(name='learner'):
    return f'Hello, {name}!'
```

```python
welcome()
```

```python
welcome('Asha')
```

## 12. Local variables and docstrings

A **local variable** is created inside a function and is used within that function call. This keeps the function's work separate from other code.

A **docstring** is a string at the start of a function body that explains what the function does. Python tools can show it as help text.

```python
def total_cost(price, quantity):
    """Return the cost of quantity items at the given price."""
    cost = price * quantity  # cost is a local variable.
    return cost
```

```python
total_cost(20, 3)
```

```python
print(total_cost.__doc__)  # Read the function documentation.
```

## 13. Built-in, user-defined, and lambda functions

**Built-in functions** come with Python, such as `len()` and `sum()`. **User-defined functions** are functions we write with `def`.

A **lambda function** is an anonymous function written with `lambda`. Anonymous means it has no name in the function definition. It can contain one expression, and that expression's value is returned automatically.

Use `def` for functions we want to name and reuse. A lambda is useful when another function needs a small rule as an argument.



The lambda has one parameter, price, and one expression, price * 0.9.

Below, `sorted()` returns a sorted list. Its `key` parameter asks for a function that supplies the sort value for each item. We sort the words by their length.

```python
words = ['banana', 'fig', 'apple']
```



The key function supplies lengths for comparison. sorted returns the original words in length order.

```python
# key calls the lambda for each word. len() returns its length.
# sorted() compares those lengths but returns the words.
sorted(words, key=lambda word: len(word))  # Lengths are 6, 3, and 5.
```

For this particular task, `sorted(words, key=len)` is shorter because `len` already is a function. The lambda shows how we can write our own rule. We do not need lambda for every small function.

## 14. Extra practice: map, filter, and list comprehensions

`map(function, items)` applies a function to each item. `filter(function, items)` keeps items for which the function returns True. Both return iterators, so we use `list()` to see the results.

These are extra ways to work on a collection. A normal `for` loop remains a clear choice.

```python
numbers = [1, 2, 3, 4]
```

```python
list(map(double, numbers))  # Apply our named function to each number.
```

![The lambda returns True for 3 and 4. filter keeps those values; list collects them into [3, 4].](attachment:lambda_filter.png)

The lambda returns True for 3 and 4. filter keeps those values; list collects them into [3, 4].

```python
# The lambda tests one number. filter() keeps values where it returns True.
kept_numbers = filter(lambda number: number > 2, numbers)
```

```python
# list() collects the kept values so we can see them.
list(kept_numbers)
```

A **list comprehension** builds a list from an expression and a loop. Read `[number * 2 for number in numbers]` as: for each number, put `number * 2` in the new list.



The for part takes one number at a time. The expression multiplies it by 2 and stores the result in a new list.

```python
# Take each number, multiply it by 2, and collect the answers.
[number * 2 for number in numbers]
```
