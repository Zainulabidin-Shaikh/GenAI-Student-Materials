## 1. Operators and operands

An **operator** is a symbol or word that performs an operation. An **operand** is a value that the operator works on. In `12 + 3`, `+` is the operator; `12` and `3` are operands.

We use arithmetic operators for calculations, comparison operators to compare values, and logical operators to combine conditions.



The whole expression 12 + 3 evaluates to 15.

## 2. Arithmetic operators

**Arithmetic operators** perform number calculations.

| Operator | Name | Example | Result |
|---|---|---|---|
| `+` | Addition | `9 + 2` | `11` |
| `-` | Subtraction | `9 - 2` | `7` |
| `*` | Multiplication | `9 * 2` | `18` |
| `/` | Division | `9 / 2` | `4.5` |
| `//` | Floor division | `9 // 2` | `4` |
| `%` | Modulo, the remainder | `9 % 2` | `1` |
| `**` | Exponentiation, a power | `3 ** 2` | `9` |

Floor division rounds the quotient down. For positive numbers it counts complete groups. Modulo gives the remainder.

```python
price = 9
```

```python
quantity = 2
```

```python
price + quantity
```

```python
price - quantity
```

```python
price * quantity  # Cost of two items.
```

```python
price / quantity
```

```python
price // quantity
```

```python
price % quantity
```

```python
3 ** 2  # 3 multiplied by 3.
```

**Operator precedence** is the order of calculation. Multiplication happens before addition. Parentheses let us choose a different order.

```python
2 + 3 * 4  # 14
```

```python
(2 + 3) * 4  # 20
```

## 3. Assignment operators

The **assignment operator** `=` binds a name to a value. An **augmented assignment operator**, such as `+=`, updates a variable using its current value.

For a number, `total += 5` has the same effect as `total = total + 5`. We use it when adding to a running total.

```python
total = 10
```

```python
total += 5  # Add 5 to the current total.
```

```python
print(total)
```

| Operator | When is the result True? |

For example, a student passes our practice rule when the score is at least 50 **and** attendance is at least 75.

## 6. Conditional statements: if, elif, and else

A **conditional statement** chooses which code to run. A **condition** is an expression Python checks as true or false.

`if` checks the first condition. `elif` means else if and checks another condition when earlier ones failed. `else` runs when none of the earlier conditions were true.

A **code block** is a group of related statements. **Indentation** is the space at the beginning of each line in a block. We use four spaces. A colon `:` starts the block.



Only one branch runs in this if/else example.

```python
if score >= 50:
    print('Pass')  # This line belongs to the if block.
```

Add `else` when we need an action for both possible results.

```python
if score >= 50:
    print('Pass')
else:
    print('Try again')
```

Use `elif` to check another range. Python runs the first matching branch, then leaves this statement.

```python
if score >= 90:
    print('Grade A')
elif score >= 70:
    print('Grade B')
else:
    print('Keep practicing')
```

With `score = 72`, the first condition is False and the second is True. The output is `Grade B`. Change the score to `95` or `40` and predict the new result.

## 8. Data structures and collections

A **data structure** is a way to organise values. Python provides several built-in collections.

| Type | Syntax | Main use | Can we change it? |
|---|---|---|---|
| List | `[10, 20]` | An ordered sequence of items | Yes |
| Tuple | `(10, 20)` | An ordered sequence whose item references stay fixed | No |
| Set | `{10, 20}` | Unique items, without index positions | Yes |
| Dictionary | `{"name": "Asha"}` | Values found by a key | Yes |

**Mutable** means an object can be changed in place. **Immutable** means it cannot. Lists are mutable; strings and tuples are immutable.

## 9. Lists: creation, indexing, and slicing

A **list** is an ordered, mutable collection. It can contain repeated values. We use it for a shopping list, student names, or scores.

List indexing and slicing work like string indexing and slicing: positions start at 0, and a slice stops before its stop position.

```python
fruits = ['apple', 'banana', 'mango']
```

```python
fruits
```

```python
fruits[0]  # First item.
```

```python
fruits[-1]  # Last item.
```

```python
fruits[0:2]  # First two items.
```

```python
len(fruits)  # Number of items.
```

## 10. List methods: add, change, and remove

A **list method** performs an action on a list. Many methods change the list itself and return `None`. Run the list in the next cell to see the change.

| Method or operation | Effect |
|---|---|
| `append(item)` | Add one item at the end |
| `extend(items)` | Add each item from another collection |
| `insert(index, item)` | Add at a chosen position |
| `list[index] = value` | Replace one item |
| `remove(value)` | Remove the first matching value |
| `pop(index)` | Remove and return the item at that position |

```python
fruits.append('orange')
```

```python
fruits
```

```python
fruits[1] = 'grapes'
```

```python
fruits
```

```python
fruits.remove('apple')
```

```python
fruits
```

```python
last_fruit = fruits.pop()  # Without an index, remove the last item.
```

```python
print(last_fruit)
```

```python
fruits
```

The **membership operators** `in` and `not in` check whether an item is present.

```python
'mango' in fruits
```

```python
'apple' not in fruits
```

## 11. Tuples

A **tuple** is an ordered collection that cannot have its items reassigned, added, or removed. We use tuples for fixed groups of values, such as a position `(x, y)`.

A tuple can contain a mutable item such as a list, but we use only numbers here. A one-item tuple needs a comma: `(5,)`.

```python
position = (3, 5)
```

```python
position[0]
```

```python
x, y = position  # Unpacking assigns each tuple item to a name.
```

```python
print(x, y)
```

## 12. Sets

A **set** stores unique items. It has no index positions, and we should not rely on its display order. Sets are useful for removing repeated labels or finding shared items.

Use `set()` for an empty set. `{}` creates an empty dictionary.

```python
topics = {'Python', 'AI', 'Python'}
```

```python
sorted(topics)  # Show the unique items in a stable order.
```

```python
topics.add('ML')  # Add one item.
```

```python
sorted(topics)
```

```python
'AI' in topics
```

**Intersection** finds items shared by two sets. **Union** combines their unique items.

```python
sorted({'AI', 'Python'} & {'Python', 'ML'})  # Intersection.
```

```python
sorted({'AI', 'Python'} | {'Python', 'ML'})  # Union.
```

## 13. Dictionaries: keys and values

A **dictionary** stores key-value pairs. A **key** is the label used to find a value. Keys are unique. Values may repeat.

In `{"name": "Asha"}`, `"name"` is the key and `"Asha"` is the value. Use a dictionary when a value needs a label, such as a student record. Dictionaries keep insertion order, but we normally look up values by key.

```python
student = {'name': 'Asha', 'score': 80}
```



Index 1 selects the second list item. The key score selects its matching dictionary value.

```python
student['name']  # Look up the value using its key.
```

```python
student['score'] = 85  # Update an existing key.
```

```python
student['city'] = 'Pune'  # Add a new key-value pair.
```

```python
student
```

`get(key, default)` lets us choose a result when a key is absent. This is useful when some student records have no city.

```python
student.get('phone', 'Not given')
```

```python
list(student.keys())  # All keys.
```

```python
list(student.values())  # All values.
```

```python
list(student.items())  # Each key with its value.
```
