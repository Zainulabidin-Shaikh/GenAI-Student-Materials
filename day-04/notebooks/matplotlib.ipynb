## 1. What is Matplotlib?

**Matplotlib** is a Python library for data visualisation. A **plot** maps data values to visible marks such as lines, bars, or points. We choose a plot type based on the question we want to answer.

`pyplot` is a Matplotlib module containing plotting functions. We import it with the common alias `plt`.

```python
import matplotlib.pyplot as plt
```

## 2. Figure, Axes, and Axis

These names have different meanings in Matplotlib.

| Term | Meaning |
|---|---|
| Figure | The whole image; it can hold several plots |
| Axes | One plotting area inside the figure |
| x-axis | The horizontal number scale or category positions |
| y-axis | The vertical number scale |
| Tick | A marked position on an axis |
| Tick label | The text or number at that tick |

The Matplotlib name **Axes** means one plotting area. It is different from the x-axis and y-axis.



The Figure is the whole image. The Axes is one plotting area inside it. The x-axis and y-axis show the two scales.

## 3. Line plots: change across an ordered sequence

A **line plot** joins points in their given order. It is useful for change over time.

The x values below are days. The y values are completed practice questions. Each x value must have a matching y value.

```python
days = [1, 2, 3, 4, 5]
```

```python
questions = [2, 4, 3, 6, 8]
```

`plt.plot(x, y)` draws the line, and `plt.show()` displays it. We keep the plotting steps together in one short cell so the output is one complete figure.

```python
plt.plot(days, questions)
plt.show()
```

## 4. Titles, axis labels, ticks, and grids

A **title** names the subject of a plot. **Axis labels** name the variables and units. A **grid** adds guide lines to help us read values.

| Function or parameter | Purpose |
|---|---|
| `title(text)` | Add a plot title |
| `xlabel(text)`, `ylabel(text)` | Name the two axes |
| `xticks(values)` | Choose positions to label on x |
| `grid(True, alpha=0.25)` | Show a light grid; alpha is opacity from 0 to 1 |

```python
plt.plot(days, questions)
plt.title('Practice questions by day')
plt.xlabel('Day')
plt.ylabel('Number of questions')
plt.xticks(days)  # Days are whole numbers.
plt.grid(True, alpha=0.25)
plt.show()
```

## 5. Line style, markers, colors, and legends

A **marker** shows each measured point. A **line style** changes the connecting line. A **legend** identifies named series.

| Parameter | Example | Meaning |
|---|---|---|
| `color` | `"teal"` | Line color |
| `marker` | `"o"` | Circle at each data point |
| `linestyle` | `"--"` | Dashed line |
| `linewidth` | `2` | Line thickness |
| `label` | `"Group A"` | Text used by the legend |

```python
group_b = [1, 3, 4, 5, 7]
```

```python
plt.plot(days, questions, color='teal', marker='o', linewidth=2, label='Group A')
plt.plot(days, group_b, color='orange', marker='s', linestyle='--', label='Group B')
plt.title('Practice by class group')
plt.xlabel('Day')
plt.ylabel('Number of questions')
plt.legend()
plt.show()
```

## 6. Bar plots: comparing categories

A **bar plot** compares values for separate categories. The height of each bar represents a value. We start the value axis at zero because bar length carries the comparison.

`plt.bar(categories, values)` draws vertical bars. `plt.barh()` draws horizontal bars.

```python
groups = ['A', 'B', 'C']
```

```python
mean_scores = [62, 74, 68]
```

```python
plt.bar(groups, mean_scores, color='teal')
plt.title('Mean score by group')
plt.xlabel('Group')
plt.ylabel('Mean score out of 100')
plt.ylim(0, 100)  # Show the full score scale.
plt.show()
```

## 7. Scatter plots: two numeric variables

A **scatter plot** shows paired numeric values. Each point is one record. Its x position gives one value and its y position gives the other.

`s` controls marker area in points squared, and `alpha` controls opacity. We use modest transparency to make overlaps easier to notice.

```python
hours = [1, 2, 3, 4, 2, 5, 3, 4]
```

```python
scores = [45, 55, 65, 78, 52, 90, 68, 74]
```

```python
plt.scatter(hours, scores, s=90, alpha=0.75, color='teal')
plt.title('Study hours and scores')
plt.xlabel('Study hours')
plt.ylabel('Score out of 100')
plt.show()
```

The points tend to rise as hours increase in this small example. This is an **association**. It does not prove that study hours alone caused the scores.

## 8. Histograms: distributions and bins

A **distribution** describes how values are spread. A **histogram** splits numeric values into intervals called **bins**, then counts the values in each bin.

A bar plot compares named categories. A histogram groups numeric values into ranges. Here the bin edges are 40, 55, 70, 85, and 100. Bins include their left edge and exclude their right edge, except the final bin includes its right edge.

```python
bin_edges = [40, 55, 70, 85, 100]
```

```python
plt.hist(scores, bins=bin_edges, color='orange', edgecolor='white')
plt.title('Score distribution')
plt.xlabel('Score interval')
plt.ylabel('Number of students')
plt.xticks(bin_edges)
plt.show()
```

For example, the first bin holds 45 and 52, so its height is 2. The bin from 55 up to 70 holds 55, 65, and 68, so its height is 3.

## 9. The Figure and Axes interface

`plt.subplots()` creates a Figure and one or more Axes. We save them as `fig` and `ax`. This makes it clear which plotting area each method acts on.

`figsize=(7, 4)` sets width and height in inches. `ax.set()` sets labels and a title together. `fig.tight_layout()` adjusts spacing to make room for labels.

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(days, questions, marker='o', color='teal')
ax.set(title='Daily practice', xlabel='Day', ylabel='Number of questions')
fig.tight_layout()
plt.show()
```

## 10. Subplots: more than one plotting area

A **subplot** is an Axes arranged beside or above another Axes in the same Figure. `plt.subplots(1, 2)` creates one row and two columns. `axes[0]` is the first plot; `axes[1]` is the second.



The Figure is the whole canvas. Each Axes holds one chart and its own title and axis labels.

```python
# 1 row and 2 columns create two Axes inside one Figure.
# axes[0] is the left chart; axes[1] is the right chart.
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot(days, questions, marker='o', color='teal')
axes[0].set(title='Daily practice', xlabel='Day', ylabel='Questions')
axes[1].bar(groups, mean_scores, color='orange')
axes[1].set(title='Group scores', xlabel='Group', ylabel='Mean score', ylim=(0, 100))
fig.tight_layout()
plt.show()
```

## 11. Saving a figure

`fig.savefig()` saves the Figure to a file. **DPI** means dots per inch; a larger value gives more pixels for the same figure size. We use 150 for a clear class image. `bbox_inches="tight"` trims spare space while keeping the labels.

```python
fig.savefig('day-04-practice-charts.png', dpi=150, bbox_inches='tight')
```
