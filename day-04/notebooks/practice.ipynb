## 1. Start with Matplotlib

**Matplotlib** is a Python library for drawing charts. A **library** is ready code we can reuse.

`pyplot` is the plotting part of Matplotlib. `plt` is its short name in our code. `import` makes it available to this notebook.

```python
import matplotlib.pyplot as plt
```

## 2. Line chart: drinks sold each day

A line chart joins points in order. It is useful for change over time.

We store five day names in a list. Their order matters because Monday comes before Tuesday.

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
```

The next list gives the sales for those same days. The first number belongs to Monday, the second to Tuesday, and so on.

```python
sales = [40, 55, 50, 70, 80]
```

The **x-axis** runs left to right. We put the days there. The **y-axis** runs bottom to top. It shows the number of drinks sold.

| Code | What it does |
| --- | --- |
| `plt.plot(days, sales)` | Draws sales in day order |
| `marker="o"` | Adds a circle at each day so we can see each recorded value |
| `plt.title(...)` | Says what the chart is about |
| `plt.xlabel(...)` | Names the horizontal axis |
| `plt.ylabel(...)` | Names the vertical axis |
| `plt.show()` | Displays the finished chart |

We keep these steps in one cell because they belong to one chart.

```python
plt.plot(days, sales, marker="o")  # Each circle marks one day.
plt.title("Drinks sold across five days")
plt.xlabel("Day")
plt.ylabel("Drinks sold")
plt.show()
```

Tuesday has 55 sales and Wednesday has 50, so the line falls between them. Friday has 80, the highest value here.

Other uses: a city's temperature each hour, or a shop's sales each month.

## 3. Bar chart: compare drink orders

A bar chart compares **categories**, which are named groups. Tea, Coffee and Juice are categories.

These are 100 orders at a cafe event. Each order is for one drink.

```python
drinks = ["Tea", "Coffee", "Juice"]
```

```python
orders = [60, 30, 10]
```

`plt.bar(drinks, orders)` gives each drink a bar. A taller bar means more orders. `color="teal"` chooses the bar color.

We add the same title, axis labels and display step used in the line chart.

```python
plt.bar(drinks, orders, color="teal")
plt.title("100 drink orders at a cafe event")
plt.xlabel("Drink")
plt.ylabel("Number of orders")
plt.show()
```

Tea has 60 orders and Coffee has 30. The difference is 30 orders.

Use a bar chart for comparing named groups, such as sales by product or students in each class. A line would suggest a sequence between the drink names.

## 4. Histogram: count marks in ranges

A histogram groups numbers into ranges and counts how many fall in each range. Each range is called a **bin**.

Here are marks for 12 practice results. Each number belongs to one student.

```python
marks = [42, 55, 61, 64, 68, 72, 74, 76, 79, 83, 88, 94]
```

We choose the edges 40, 60, 80 and 100. These four edges make three ranges.

40 up to 60 contains 42 and 55. A mark of 60 starts the next range. The final range includes 100.

```python
bin_edges = [40, 60, 80, 100]
```

`plt.hist(marks, bins=bin_edges)` counts the marks in our chosen ranges.

In `bins=bin_edges`, `bins` is the setting the function expects. `bin_edges` is our list of boundaries.

`edgecolor="white"` adds a white border so we can see where one bin ends and the next begins.

```python
plt.hist(marks, bins=bin_edges, edgecolor="white")
plt.title("Marks from 12 practice results")
plt.xlabel("Marks out of 100")
plt.ylabel("Number of students")
plt.show()
```

The three bar heights are **2, 7 and 3 students**. Most marks are from 60 up to 80.

**Bar chart:** compares named groups such as Tea and Coffee.

**Histogram:** counts numbers in ranges, such as marks or delivery times.

## 5. Pie chart: parts of all orders

We reuse `drinks` and `orders` from the bar chart. Their total is 100.

`plt.pie(orders)` draws one slice per value. `labels=drinks` puts a drink name beside each slice. A pie chart has no x-axis or y-axis, so we only add a title and show it.

```python
plt.pie(orders, labels=drinks)
plt.title("All 100 orders make one whole")
plt.show()
```

Tea has 60 of the 100 orders, so it takes 60% of the pie. Coffee takes 30% and Juice takes 10%.

Use a pie for a few parts of one total, such as how a monthly budget is split. Use bars if two amounts are close and you want to compare them clearly.

## 6. Scatter plot: one dot for each student

A **scatter plot** shows two numbers for each record. Here one dot is one student. Its position shows study hours and marks.

The first hour value and the first mark belong to the same student. The next pair belongs to the next student.

```python
hours = [1, 2, 3, 4, 5]
```

```python
scores = [45, 55, 60, 75, 85]
```

`plt.scatter(hours, scores)` draws separate dots. We leave them unconnected because each dot belongs to a different student.

Use a scatter plot for paired numbers, such as a house's size and price, or study hours and marks.

```python
plt.scatter(hours, scores)
plt.title("Study hours and marks")
plt.xlabel("Study hours")
plt.ylabel("Marks out of 100")
plt.show()
```

**What we see:** The dots rise from left to right. Students with more study hours have higher marks in this small practice example. Other things can also affect marks.

A line chart joins values in a sequence, such as days. A scatter plot keeps each student's pair as a separate dot.

## 7. Draw the same charts with Seaborn

**Seaborn** is another plotting library. It uses Matplotlib to draw. For example, Seaborn can count students in each class straight from a class table and draw their bars.

You can use Matplotlib alone. Here we try Seaborn with the same lists so only the drawing function changes. `sns` is the short name we use for Seaborn.

```python
import seaborn as sns
```

`sns.barplot(x=drinks, y=orders)` puts drink names on x and order totals on y. Each drink has one value here, so its bar shows that value.

Seaborn uses Matplotlib underneath. That is why `plt.title()`, `plt.xlabel()`, `plt.ylabel()` and `plt.show()` still work. The same `color="teal"` setting keeps the bars teal.

```python
sns.barplot(x=drinks, y=orders, color="teal")
plt.title("100 drink orders at a cafe event")
plt.xlabel("Drink")
plt.ylabel("Number of orders")
plt.show()
```

**What we see:** The Seaborn bars still show 60, 30 and 10 orders. Both bar charts answer the same question about drink choices.

### The same line chart

`sns.lineplot(x=days, y=sales)` uses our day names and daily totals. `marker="o"` marks each recorded day, just as it did in Matplotlib.

```python
sns.lineplot(x=days, y=sales, marker="o")
plt.title("Drinks sold across five days")
plt.xlabel("Day")
plt.ylabel("Drinks sold")
plt.show()
```

**What we see:** Wednesday still has the small dip, and Friday still has the highest sales. The data and the day order stayed the same.

### The same histogram

`sns.histplot()` counts numbers in ranges. `x=marks` chooses the numbers. `bins=bin_edges` keeps the same boundaries so we can compare it with the Matplotlib chart.

```python
sns.histplot(x=marks, bins=bin_edges)
plt.title("Marks from 12 practice results")
plt.xlabel("Marks out of 100")
plt.ylabel("Number of students")
plt.show()
```

**What we see:** The range counts are still 2, 7 and 3. Most marks fall from 60 up to 80.

### The same scatter plot

`sns.scatterplot()` draws one dot per pair. `x=hours` chooses horizontal positions. `y=scores` chooses vertical positions.

```python
sns.scatterplot(x=hours, y=scores)
plt.title("Study hours and marks")
plt.xlabel("Study hours")
plt.ylabel("Marks out of 100")
plt.show()
```

**What we see:** The five dots have the same positions as before. The upward pattern comes from the values in our two lists.

### A pie with Seaborn colors

Seaborn has no pie-chart function. We use Matplotlib's `plt.pie()` to draw it.

A **palette** is a list of colors. `sns.color_palette("pastel", 3)` gives us three light colors, one for each drink. We save that list before using it.

```python
pie_colors = sns.color_palette("pastel", 3)
```

`colors=pie_colors` gives those colors to Matplotlib. This shows how the two libraries can work together.

```python
plt.pie(orders, labels=drinks, colors=pie_colors)
plt.title("All 100 orders make one whole")
plt.show()
```

**What we see:** Only the colors changed. Tea still takes 60% of the whole, Coffee takes 30% and Juice takes 10%.

## 8. Try a small flower dataset

The **Iris dataset** is a table of flower measurements. Each row is one flower. It has 150 rows and three flower types: setosa, versicolor and virginica.

- **Species** means the flower type.
- **Petal** is one of the flower's colored parts.
- **Sepal** is a leaf-like part around the flower.
- **Length** measures how long a part is. **Width** measures how wide it is.

The length and width values are in centimeters, written as **cm**.

Seaborn's sample-data collection includes this table. We keep a local copy for class. `pd` is the short name for pandas, the table library from Day 03.

```python
import pandas as pd
```

`pd.read_csv()` reads a table from a CSV file. `..` means go up from the notebooks folder to the Day 04 folder. Then `assets/iris.csv` selects our saved file.

```python
iris = pd.read_csv("../assets/iris.csv")
```

`head()` shows the first five rows. Look at the column names before choosing a chart.

```python
iris.head()
```

Each row has four measurements and one species name. We will use `petal_length`, `petal_width` and `species` below.

### Flower counts: how many of each type?

`sns.countplot()` counts rows in each category. `data=iris` chooses our table. `x="species"` chooses the column with flower names. There is no `y` input because Seaborn counts the rows for us.

```python
sns.countplot(data=iris, x="species", color="teal")
plt.title("Flowers of each species")
plt.xlabel("Species")
plt.ylabel("Number of flowers")
plt.show()
```

**What we see:** All three bars reach 50. Each species has the same number of flowers in this table. This is different from our cafe orders, where one drink was much more common.

### Petal lengths: where do the values fall?

We use a histogram because petal length is a number. `x="petal_length"` chooses that column. `bins=10` splits the full length range into ten equal bins. This gives more detail than the three bins in our marks example.

```python
sns.histplot(data=iris, x="petal_length", bins=10)
plt.title("Petal lengths in the Iris dataset")
plt.xlabel("Petal length (cm)")
plt.ylabel("Number of flowers")
plt.show()
```

**What we see:** Many petals are shorter than 2 cm. The longer petals sit in another group, from about 3 cm upward. This chart shows lengths, but it does not tell us which species each flower belongs to.

### Petal length and width: show the species

A scatter plot can show length and width together. Each dot is one flower.

`hue="species"` uses a different color for each flower type. **Hue** is Seaborn's setting for color groups. The **legend** matches each color to a species name.

```python
sns.scatterplot(data=iris, x="petal_length", y="petal_width", hue="species")
plt.title("Petal length and width by species")
plt.xlabel("Petal length (cm)")
plt.ylabel("Petal width (cm)")
plt.show()
```

**What we see:** Setosa has short, narrow petals and sits apart at the lower left. Versicolor and virginica have longer petals, with some overlap between them. These two measurements do not separate every flower type perfectly.

The dots mostly rise as we move right: longer petals also tend to be wider in this dataset.

## Choose a chart

| We want to see... | Choose... |
| --- | --- |
| Change across days or months | Line |
| Which named group has more | Bar |
| How numbers fall into ranges | Histogram |
| Each part's share of one whole | Pie |
| How two numbers vary together | Scatter |
